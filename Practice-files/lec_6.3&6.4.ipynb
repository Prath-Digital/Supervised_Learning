
# Evaluation Metrics for Classification


## 1. Import Libraries

```python

import pandas as pd
import numpy as np

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    roc_curve,
    roc_auc_score
)

import matplotlib.pyplot as plt
```


## 2. Create Example Data

y_true:
Actual values

y_pred:
Predicted values


```python

y_true = [1, 1, 1, 0, 0, 0]

y_pred = [1, 1, 0, 1, 0, 0]

print("Actual Values :", y_true)
print("Predicted Values:", y_pred)
```


## 3. Confusion Matrix

Structure:

|                | Predicted Positive | Predicted Negative |
|----------------|-------------------|-------------------|
| Actual Positive | TP | FN |
| Actual Negative | FP | TN |


```python

cm = confusion_matrix(
    y_true,
    y_pred
)

print("Confusion Matrix")
print(cm)
```


## 4. Extract TP, TN, FP, FN

Confusion Matrix Output:

[[TN FP]
 [FN TP]]


```python

TN, FP, FN, TP = confusion_matrix(
    y_true,
    y_pred
).ravel()

print("True Positive :", TP)
print("True Negative :", TN)
print("False Positive:", FP)
print("False Negative:", FN)
```


## 5. Accuracy

Accuracy measures overall correctness.


```python

accuracy = accuracy_score(
    y_true,
    y_pred
)

print("Accuracy:", accuracy)
```


## 6. Precision

Precision answers:

Out of predicted positives,
how many were actually positive?


```python

precision = precision_score(
    y_true,
    y_pred
)

print("Precision:", precision)
```


## 7. Recall

Recall answers:

Out of actual positives,
how many did the model correctly identify?


```python

recall = recall_score(
    y_true,
    y_pred
)

print("Recall:", recall)
```


## 8. F1 Score

F1 Score balances:
- Precision
- Recall


```python

f1 = f1_score(
    y_true,
    y_pred
)

print("F1 Score:", f1)
```


## 9. Classification Report


```python

report = classification_report(
    y_true,
    y_pred
)

print(report)
```


## 10. Predicted Probabilities

ROC-AUC uses probabilities.


```python

y_prob = [0.9, 0.8, 0.4, 0.7, 0.2, 0.1]

print(y_prob)
```


## 11. ROC-AUC Score


```python

auc = roc_auc_score(
    y_true,
    y_prob
)

print("ROC-AUC Score:", auc)
```


## 12. ROC Curve


```python

fpr, tpr, thresholds = roc_curve(
    y_true,
    y_prob
)

plt.figure(figsize=(6,5))

plt.plot(fpr, tpr)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.show()
```


## 13. Imbalanced Dataset Problem

Accuracy can be misleading in imbalanced datasets.



## 14. When to Use Which Metric

| Situation | Best Metric |
|---|---|
| Balanced dataset | Accuracy |
| Fraud detection | Precision |
| Disease detection | Recall |
| Imbalanced dataset | F1 Score |
| Probability ranking | ROC-AUC |



## 15. Manual Metric Calculations


```python

TP = 2
TN = 2
FP = 1
FN = 1

accuracy_manual = (TP + TN) / (TP + TN + FP + FN)

precision_manual = TP / (TP + FP)

recall_manual = TP / (TP + FN)

f1_manual = 2 * (
    precision_manual * recall_manual
) / (
    precision_manual + recall_manual
)

print("Manual Accuracy :", accuracy_manual)
print("Manual Precision:", precision_manual)
print("Manual Recall   :", recall_manual)
print("Manual F1 Score :", f1_manual)
```


## 16. Final Summary

Important Metrics:

- Confusion Matrix
- Accuracy
- Precision
- Recall
- F1 Score
- ROC-AUC

These metrics help evaluate classification models correctly.
