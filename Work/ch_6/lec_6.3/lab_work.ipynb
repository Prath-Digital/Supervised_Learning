y_true = [1, 0, 1, 1, 0, 0, 1, 0, 1, 0]

````python
import numpy as np
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, roc_curve, auc
import matplotlib.pyplot as plt

y_true = np.array([1, 0, 1, 1, 0, 0, 1, 0, 1, 0])
y_pred = np.array([1, 0, 1, 0, 1, 0, 1, 0, 0, 0])
y_proba = np.array([0.92, 0.10, 0.87, 0.45, 0.18, 0.72, 0.95, 0.30, 0.40, 0.05])

print('Dataset loaded')
````

## Q1: Accuracy & Confusion Matrix

### 01 Definitions
- **True Positive (TP)**: Model correctly predicts Positive when actual is Positive.
- **False Positive (FP)**: Model incorrectly predicts Positive when actual is Negative.
- **True Negative (TN)**: Model correctly predicts Negative when actual is Negative.
- **False Negative (FN)**: Model incorrectly predicts Negative when actual is Positive.

````python
# Q1.02 - Classify each sample
for i in range(len(y_true)):
    actual = 'Positive' if y_true[i] == 1 else 'Negative'
    pred = 'Positive' if y_pred[i] == 1 else 'Negative'
    print(f'Sample {i}: Actual={actual}, Pred={pred}')
````

````python
# Q1.03 Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
print('Confusion Matrix:')
print(cm)

tn, fp, fn, tp = cm.ravel()
print(f'TP={tp}, FP={fp}, TN={tn}, FN={fn}')
````

````python
# Q1.04 Accuracy
accuracy = (tp + tn) / len(y_true)
print(f'Accuracy: {accuracy:.2f} or {accuracy*100:.0f}%')
````

````python
# Q1.05 Error Rate
error_rate = 1 - accuracy
print(f'Error Rate: {error_rate:.2f}')
````

````python
# Q1.06 All-zero predictor
y_pred_zero = np.zeros(10, dtype=int)
acc_zero = accuracy_score(y_true, y_pred_zero)
print(f'All-zero accuracy: {acc_zero}')
````

## Q2: Type I & Type II Errors

### Definitions
- **Type I Error (False Positive)**: Rejecting a true null hypothesis. In medicine: diagnosing healthy as diseased. In law: convicting innocent. In spam: marking good email as spam.
- **Type II Error (False Negative)**: Failing to reject a false null hypothesis. In medicine: missing a disease. In law: acquitting guilty. In spam: letting spam into inbox.

````python
# Q2.03 From CM
print(f'Type I Errors (FP): {fp}')
print(f'Type II Errors (FN): {fn}')
````

````python
# Rates
fpr = fp / (fp + tn)
fnr = fn / (fn + tp)
print(f'FPR: {fpr:.2f}, FNR: {fnr:.2f}')
````

## Q3: Precision, Recall & F1 Score

````python
# Precision
precision = tp / (tp + fp)
print(f'Precision: {precision:.2f}')
````

````python
# Recall
recall = tp / (tp + fn)
print(f'Recall: {recall:.2f}')
````

````python
# Specificity
specificity = tn / (tn + fp)
print(f'Specificity: {specificity:.2f}')
````

````python
# F1 Score
f1 = 2 * (precision * recall) / (precision + recall)
print(f'F1: {f1:.2f}')
````

## Q4: TPR, FPR & ROC Curve

````python
# At threshold 0.5
print('At threshold 0.5: TPR (Recall) =', recall, 'FPR =', fpr)
````

````python
# ROC Curve manual
fpr_list, tpr_list, thresholds = roc_curve(y_true, y_proba)
print('ROC points ready for plotting')
````

````python
# Plot ROC
plt.figure(figsize=(8,6))
plt.plot(fpr_list, tpr_list, marker='o')
plt.plot([0,1], [0,1], 'r--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.grid(True)
plt.show()
````

````python
# AUC
roc_auc = auc(fpr_list, tpr_list)
print(f'AUC: {roc_auc:.2f}')
````
