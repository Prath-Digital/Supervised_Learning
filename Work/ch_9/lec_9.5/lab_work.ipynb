# Q.1 Conditional Probability


### 01. Definition of Conditional Probability

**Question:**  
Define conditional probability: $P(A|B) = \frac{P(A \cap B)}{P(B)}$, read as "the probability of A given that B has occurred."  
Explain intuitively — when you observe B, you shrink the sample space from $\Omega$ to B; $P(A|B)$ asks what fraction of B contains A.  
Verify that $P(A|B) \neq P(B|A)$ with a concrete example.


**Answer:**

**Definition**  
The **conditional probability** of event $A$ given that event $B$ has occurred is

$$
P(A|B) = \frac{P(A \cap B)}{P(B)}, \quad \text{provided } P(B) > 0.
$$

**Intuitive Explanation**  
- The original sample space is $\Omega$.  
- Observing that $B$ occurred means we discard every outcome outside $B$. The new sample space is just $B$.  
- Among the outcomes that remain (those in $B$), the proportion that also belong to $A$ is exactly $\frac{|A \cap B|}{|B|}$ (in the finite equally-likely case), or more generally $\frac{P(A \cap B)}{P(B)}$.  
Thus $P(A|B)$ measures “how much of $B$ is occupied by $A$.”

**Concrete Counter-example showing $P(A|B) \neq P(B|A)$**  

Consider a standard deck of 52 cards.  
Let  
- $A$ = “card is an Ace”  → $P(A) = 4/52 = 1/13$  
- $B$ = “card is a Spade” → $P(B) = 13/52 = 1/4$  
- $A \cap B$ = “Ace of Spades” → $P(A \cap B) = 1/52$

Then

$$
P(A|B) = \frac{1/52}{1/4} = \frac{1}{13}, \qquad
P(B|A) = \frac{1/52}{1/13} = \frac{1}{4}.
$$

Clearly $\frac{1}{13} \neq \frac{1}{4}$.  
(Being a Spade does not change the chance of being an Ace, but being an Ace does change the chance of being a Spade.)


### 02. Manual Computation – Bag of Balls

**Question:**  
A bag has 5 red balls and 3 blue balls. You draw one ball without replacement.  
Event $A$ = second ball is red, Event $B$ = first ball is red.  
Compute $P(A|B)$, $P(A|B^c)$, and $P(A)$.  
Show all working using the definition of conditional probability.


**Answer:**

Total balls = 8 (5 red + 3 blue).

**1. $P(A|B)$ – probability second is red given first was red**

After drawing a red ball, 4 red and 3 blue remain → 7 balls left.  
$$
P(A|B) = \frac{4}{7}.
$$
By definition:
$$
P(A \cap B) = P(\text{1st red and 2nd red}) = \frac{5}{8} \cdot \frac{4}{7} = \frac{20}{56} = \frac{5}{14},
$$
$$
P(B) = \frac{5}{8},
$$
$$
P(A|B) = \frac{P(A\cap B)}{P(B)} = \frac{5/14}{5/8} = \frac{5}{14} \cdot \frac{8}{5} = \frac{4}{7}.
$$

**2. $P(A|B^c)$ – probability second is red given first was blue**

After drawing a blue ball, 5 red and 2 blue remain → 7 balls.  
$$
P(A|B^c) = \frac{5}{7}.
$$
Check with definition:
$$
P(A \cap B^c) = P(\text{1st blue and 2nd red}) = \frac{3}{8} \cdot \frac{5}{7} = \frac{15}{56},
$$
$$
P(B^c) = \frac{3}{8},
$$
$$
P(A|B^c) = \frac{15/56}{3/8} = \frac{15}{56} \cdot \frac{8}{3} = \frac{5}{7}.
$$

**3. $P(A)$ – unconditional probability that the second ball is red**

By the Law of Total Probability:
$$
\begin{align*}
P(A) &= P(A|B)P(B) + P(A|B^c)P(B^c) \\
&= \frac{4}{7}\cdot\frac{5}{8} + \frac{5}{7}\cdot\frac{3}{8} \\
&= \frac{20}{56} + \frac{15}{56} = \frac{35}{56} = \frac{5}{8}.
\end{align*}
$$
(As expected: by symmetry the second ball is equally likely to be any of the 8 balls, so $P(\text{red}) = 5/8$.)


### 03. Chain Rule of Probability

**Question:**  
Chain rule of probability:  
$$
P(A \cap B) = P(A|B)\cdot P(B) = P(B|A)\cdot P(A).
$$
Extend to three events:  
$$
P(A \cap B \cap C) = P(A|B\cap C)\cdot P(B|C)\cdot P(C).
$$  
Use the chain rule to compute  
$P(\text{drawing 3 red balls in a row without replacement})$ from a bag of 5 red and 3 blue.


**Answer:**

**Chain Rule (two events)**  
$$
P(A\cap B) = P(A|B)\,P(B) = P(B|A)\,P(A).
$$

**Chain Rule (three events)**  
$$
\begin{align*}
P(A\cap B\cap C)
&= P(A|B\cap C)\,P(B\cap C) \\
&= P(A|B\cap C)\,P(B|C)\,P(C).
\end{align*}
$$
(The order of conditioning can be permuted arbitrarily.)

**Application – three red balls without replacement**

Let $R_i$ = “$i$-th ball drawn is red.”  
We want $P(R_1\cap R_2\cap R_3)$.

$$
\begin{align*}
P(R_1\cap R_2\cap R_3)
&= P(R_3|R_1\cap R_2)\,P(R_2|R_1)\,P(R_1) \\
&= \frac{3}{6}\cdot\frac{4}{7}\cdot\frac{5}{8} \\
&= \frac{3}{6}\cdot\frac{4}{7}\cdot\frac{5}{8} = \frac{60}{336} = \frac{5}{28}.
\end{align*}
$$

(After two reds are removed, 3 reds and 3 blues remain, so the last factor is $3/6$.)


### 04. Statistical Independence

**Question:**  
Define statistical independence: $A$ and $B$ are independent iff $P(A\cap B)=P(A)\cdot P(B)$,  
equivalently $P(A|B)=P(A)$.  

Check independence: in a fair die roll, $A=\{\text{even}\}$ and $B=\{ \le 3\}$.  
Compute $P(A)$, $P(B)$, $P(A\cap B)$, and test independence.  

Repeat for $A=\{\text{even}\}$ and $B=\{1,2,3,4\}$.


**Answer:**

**Definition**  
Events $A$ and $B$ are **(statistically) independent** if and only if
$$
P(A\cap B) = P(A)\,P(B)
$$
(or equivalently $P(A|B)=P(A)$ whenever $P(B)>0$).

**Fair six-sided die**

Sample space $\Omega=\{1,2,3,4,5,6\}$, each outcome probability $1/6$.

**Case 1**  
$A = \{\text{even}\} = \{2,4,6\}$ → $P(A)=3/6=1/2$  
$B = \{\le 3\} = \{1,2,3\}$ → $P(B)=3/6=1/2$  
$A\cap B = \{2\}$ → $P(A\cap B)=1/6$

$$
P(A)P(B) = \frac12\cdot\frac12 = \frac14 \neq \frac16 = P(A\cap B).
$$
Hence **dependent**.

**Case 2**  
$A = \{\text{even}\} = \{2,4,6\}$ → $P(A)=1/2$  
$B = \{1,2,3,4\}$ → $P(B)=4/6=2/3$  
$A\cap B = \{2,4\}$ → $P(A\cap B)=2/6=1/3$

$$
P(A)P(B) = \frac12\cdot\frac23 = \frac13 = P(A\cap B).
$$
Hence **independent**.


### 05. Law of Total Probability

**Question:**  
Law of Total Probability:

$$
P(A) = \sum_k P(A|B_k)\,P(B_k)
$$

where $\{B_1,\dots,B_n\}$ is a partition of $\Omega$.

Medical example: a disease affects 1 % of the population. A test is 95 % sensitive  
($P(+|\text{Disease})=0.95$) and 90 % specific ($P(-|\text{No Disease})=0.90$).  
Compute $P(+)$ — the overall probability of a positive test result.


**Answer:**

**Law of Total Probability**  
If $\{B_k\}$ is a countable partition of the sample space (mutually exclusive and exhaustive), then for any event $A$
$$
P(A) = \sum_k P(A|B_k)\,P(B_k).
$$

**Medical Example**

Let  
- $D$ = has disease, $P(D)=0.01$, $P(D^c)=0.99$  
- $+$ = positive test  

Given:  
$$
P(+|D)=0.95, \qquad P(+|D^c)=1-0.90=0.10.
$$

$$
\begin{align*}
P(+)
&= P(+|D)P(D) + P(+|D^c)P(D^c) \\
&= (0.95)(0.01) + (0.10)(0.99) \\
&= 0.0095 + 0.099 = 0.1085.
\end{align*}
$$

So roughly **10.85 %** of all people will test positive.


### 06. Conditional Independence vs Marginal Independence

**Question:**  
Explain conditional independence: $A$ and $B$ are conditionally independent given $C$ iff  
$$
P(A\cap B|C) = P(A|C)\,P(B|C).
$$
This does **NOT** mean $A$ and $B$ are marginally independent.

Construct a concrete example where two events are conditionally independent given a third but are marginally dependent  
(hint: use a common cause / confounding variable).


**Answer:**

**Conditional Independence**  
$A \perp B \mid C$ means
$$
P(A\cap B \mid C) = P(A|C)\,P(B|C)
$$
(or equivalently $P(A|B,C)=P(A|C)$).

**Important distinction**  
Conditional independence does **not** imply ordinary (marginal) independence, and vice versa.

**Classic concrete example – common cause**

Imagine two fair coins and a biased coin.  
We first flip a fair “selector” coin $C$:  
- If $C=\text{Heads}$ we use a fair coin for both $A$ and $B$.  
- If $C=\text{Tails}$ we use a two-headed coin for both $A$ and $B$.

More formally, let the sample space be generated as follows:

1. Flip a fair coin $C$ (the common cause).  
2. Given $C$, generate two independent Bernoulli random variables $A$ and $B$ with the same bias:  
   - $P(A=1|C=\text{H})=P(B=1|C=\text{H})=1/2$  
   - $P(A=1|C=\text{T})=P(B=1|C=\text{T})=1$  

Then, **conditionally on $C$**, $A$ and $B$ are independent by construction.  

But **marginally** they are dependent:  
$$
\begin{align*}
P(A=1) &= \tfrac12\cdot\tfrac12 + \tfrac12\cdot 1 = \tfrac34, \\
P(B=1) &= \tfrac34, \\
P(A=1,B=1) &= \tfrac12\cdot\tfrac14 + \tfrac12\cdot 1 = \tfrac58 \neq \tfrac34\cdot\tfrac34 = \tfrac9{16}.
\end{align*}
$$
Hence $A$ and $B$ are **marginally dependent** (they both tend to be 1 when the two-headed coin is chosen) yet **conditionally independent given $C$**.

# Q.2 Bayes' Theorem


### 01. Derivation of Bayes' Theorem

**Question:**  
Derive Bayes' theorem from the definition of conditional probability:  
$$
P(A|B) = \frac{P(B|A)\,P(A)}{P(B)}.
$$
Name each term precisely:  
- $P(A)$ = prior (belief before seeing evidence)  
- $P(B|A)$ = likelihood (how probable the evidence is under hypothesis $A$)  
- $P(B)$ = marginal likelihood (evidence probability)  
- $P(A|B)$ = posterior (updated belief).


**Answer:**

**Derivation**

By definition of conditional probability:
$$
P(A|B) = \frac{P(A\cap B)}{P(B)}, \qquad
P(B|A) = \frac{P(A\cap B)}{P(A)}.
$$

From the second equation, $P(A\cap B) = P(B|A)\,P(A)$.  
Substitute into the first:
$$
P(A|B) = \frac{P(B|A)\,P(A)}{P(B)}.
$$

This is **Bayes' theorem**.

**Terminology**

| Symbol          | Name                     | Meaning                                      |
|-----------------|--------------------------|----------------------------------------------|
| $P(A)$      | **Prior**                | Belief in hypothesis $A$ before evidence |
| $P(B|A)$    | **Likelihood**           | How well hypothesis $A$ predicts data $B$ |
| $P(B)$      | **Marginal likelihood** / Evidence | Total probability of observing $B$ |
| $P(A|B)$    | **Posterior**            | Updated belief in $A$ after seeing $B$ |

The denominator can be expanded by the law of total probability:
$$
P(B) = \sum_i P(B|A_i)\,P(A_i)
$$
when $\{A_i\}$ partitions the space.


### 02. Medical Diagnosis Example

**Question:**  
Medical diagnosis example:  
$P(\text{Disease})=0.01$, $P(+|\text{Disease})=0.95$, $P(+|\text{No Disease})=0.10$.  
Compute $P(\text{Disease}|+)$ using Bayes' theorem and the Law of Total Probability for the denominator.  
Show every step. Interpret the result — why is $P(\text{Disease}|+)$ so much lower than $P(+|\text{Disease})=0.95$, even for a sensitive test?


**Answer:**

**Given**
$$
P(D)=0.01,\quad P(D^c)=0.99,
$$
$$
P(+|D)=0.95,\quad P(+|D^c)=0.10.
$$

**Bayes' theorem**
$$
P(D|+) = \frac{P(+|D)\,P(D)}{P(+)}.
$$

**Denominator via Law of Total Probability**
$$
\begin{align*}
P(+)
&= P(+|D)P(D) + P(+|D^c)P(D^c) \\
&= (0.95)(0.01) + (0.10)(0.99) \\
&= 0.0095 + 0.099 = 0.1085.
\end{align*}
$$

**Posterior**
$$
P(D|+) = \frac{0.95\times 0.01}{0.1085} = \frac{0.0095}{0.1085} \approx 0.0876 \approx 8.76\,\%.
$$

**Interpretation – the base-rate fallacy**

Even though the test is 95 % sensitive, the disease is extremely rare (only 1 %).  
Most positive tests therefore come from the large healthy population (false positives).  
Out of 10 000 people:
- 100 truly have the disease → ~95 true positives  
- 9 900 healthy → ~990 false positives  

Total positives ≈ 1 085, of which only 95 are true → ≈ 8.8 %.  
The low prior (base rate) dominates the moderately high likelihood.


### 03. Sequential Updating

**Question:**  
Sequential updating: you test the same patient twice with independent tests.  
Test 1 is positive: update $P(\text{Disease})$ using Bayes' theorem to get posterior.  
Use posterior as the new prior and apply Bayes' theorem for a second positive test.  
Show that sequential updating equals a single update on both tests simultaneously.


**Answer:**

Assume the two tests are conditionally independent given disease status, with the same parameters as before:
$$
P(+|D)=0.95,\quad P(+|D^c)=0.10.
$$

**Step 1 – first positive test**
$$
P(D|+_1) = \frac{0.95\times 0.01}{0.1085} \approx 0.0876.
$$
Call this $p_1 = 0.0876$.

**Step 2 – second positive test, using $p_1$ as prior**
$$
\begin{align*}
P(+) &= 0.95\,p_1 + 0.10\,(1-p_1) \approx 0.95\times0.0876 + 0.10\times0.9124 \\
&= 0.08322 + 0.09124 = 0.17446, \\
P(D|+_1,+_2) &= \frac{0.95\times0.0876}{0.17446} \approx 0.477.
\end{align*}
$$

**Simultaneous update on both tests**

Likelihood of two independent positives:
$$
P(+_1,+_2|D) = 0.95^2 = 0.9025,
$$
$$
P(+_1,+_2|D^c) = 0.10^2 = 0.01.
$$

$$
\begin{align*}
P(+_1,+_2)
&= 0.9025\times0.01 + 0.01\times0.99 = 0.009025 + 0.0099 = 0.018925, \\
P(D|+_1,+_2)
&= \frac{0.9025\times0.01}{0.018925} \approx 0.4769.
\end{align*}
$$

The two results match (within rounding error).  
**Sequential Bayesian updating is equivalent to a single joint update** when the tests are conditionally independent given the hypothesis.

### 05. Prior Sensitivity / False-Positive Paradox

**Question:**  
Prior sensitivity: repeat the medical diagnosis from task 02 with $P(\text{Disease})=0.001$ (rare disease).  
How does $P(\text{Disease}|+)$ change? Explain qualitatively why the base rate (prior) dominates the posterior even for highly sensitive tests — this is the false-positive paradox.


**Answer:**

**New prior** $P(D)=0.001$, $P(D^c)=0.999$.  
Sensitivities unchanged: $P(+|D)=0.95$, $P(+|D^c)=0.10$.

$$
\begin{align*}
P(+) &= 0.95\times0.001 + 0.10\times0.999 = 0.00095 + 0.0999 = 0.10085, \\
P(D|+) &= \frac{0.00095}{0.10085} \approx 0.00942 \approx 0.94\,\%.
\end{align*}
$$

**Qualitative explanation – the false-positive paradox**

When the disease is extremely rare, the absolute number of false positives from the large healthy population overwhelms the true positives.  
Even a test that is 95 % sensitive and 90 % specific produces a posterior probability of disease of less than 1 % after a positive result.  

This is why screening tests for rare conditions are almost always followed by a more specific confirmatory test: a single positive is still weak evidence when the base rate is tiny.


### 06. Bayes' Theorem for Continuous Variables

**Question:**  
Bayes' theorem for continuous variables: replace $P(B|A)$ with a likelihood function $f(x|\theta)$ and $P(A)$ with a prior density $\pi(\theta)$.  
The posterior is  
$$
\pi(\theta|x) \propto f(x|\theta)\,\pi(\theta).
$$
Explain why the normalising constant $f(x)$ is often intractable and why MCMC methods exist to sample from the posterior without computing it.


**Answer:**

**Continuous Bayes**

For a continuous parameter $\theta$ and observed data $x$,
$$
\pi(\theta|x) = \frac{f(x|\theta)\,\pi(\theta)}{f(x)},
$$
where the marginal likelihood (evidence) is the integral
$$
f(x) = \int f(x|\theta)\,\pi(\theta)\,d\theta.
$$

In practice we usually work with the un-normalised posterior
$$
\pi(\theta|x) \propto f(x|\theta)\,\pi(\theta).
$$

**Why the normaliser is intractable**

- The integral $\int f(x|\theta)\pi(\theta)\,d\theta$ rarely has a closed form (except for conjugate pairs).  
- In high dimensions the integral becomes a high-dimensional integral that cannot be evaluated by ordinary quadrature.  
- Even if a numerical approximation exists, it must be recomputed for every new data set.

**MCMC solution**

Markov-Chain Monte-Carlo methods (Metropolis–Hastings, Gibbs sampling, Hamiltonian Monte Carlo, …) construct a Markov chain whose **stationary distribution** is exactly the target posterior $\pi(\theta|x)$.  

Because the acceptance ratio of Metropolis–Hastings only involves the **ratio** of posterior densities, the unknown normalising constant $f(x)$ cancels out and never needs to be computed.  
After a burn-in period the samples can be treated as approximate draws from the posterior, allowing us to estimate any posterior expectation of interest.

# Q.3 Naive Bayes Classifier


### 01. Definition of the Naive Bayes Classifier

**Question:**  
Define the Naive Bayes classifier: given feature vector $\mathbf{x}=(x_1,\dots,x_d)$ and class $y$,  
predict

$$
\hat y = \arg\max_y\, P(y|\mathbf{x}) = \arg\max_y\, P(y)\prod_{i=1}^d P(x_i|y),
$$

where the naive assumption is

$$
P(x_1,\dots,x_d|y) = \prod_{i=1}^d P(x_i|y).
$$

Explain why this assumption is called “naive” — features are almost never truly independent given the class, which is almost never true in practice.


**Answer:**

**Definition**

The **Naive Bayes classifier** chooses the class label that maximises the posterior probability under the **conditional independence assumption**:

$$
\hat y = \arg\max_y\, P(y|\mathbf{x})
= \arg\max_y\, \frac{P(y)\,P(\mathbf{x}|y)}{P(\mathbf{x})}
= \arg\max_y\, P(y)\prod_{i=1}^d P(x_i|y).
$$
(The denominator $P(\mathbf{x})$ does not depend on $y$ and can be ignored for maximisation.)

**Why “naive”?**

The assumption
$$
P(x_1,\dots,x_d|y) = \prod_i P(x_i|y)
$$
states that, **given the class**, every feature is independent of every other feature.  

In almost every real data set this is false:  
- words in a document are correlated,  
- pixels in an image are spatially dependent,  
- medical symptoms tend to co-occur, etc.  

Nevertheless the model often works surprisingly well (see next question).


### 02. Why Naive Bayes Still Works

**Question:**  
Explain why Naive Bayes works despite the violated independence assumption:  
even if the independence assumption is wrong, Naive Bayes only needs to **rank classes correctly**, not produce exact posteriors.  
The $\arg\max$ over $y$ is often correct even when the probability estimates are badly calibrated.  
This is why it beats more complex models on many real datasets.


**Answer:**

Naive Bayes is a **discriminative ranking** method rather than a calibrated probability estimator.

- Classification only requires
  $$
  P(y=1|\mathbf{x}) > P(y=0|\mathbf{x})
  $$
  (or the analogous inequality for multi-class).  
- Even when the absolute values of the estimated posteriors are far from the true probabilities (because of the independence assumption), the **relative ordering** of the classes is frequently preserved.  
- Consequently the decision boundary can still be close to optimal.

Empirically this “wrong model, right ranking” phenomenon is observed on text classification, spam filtering, and many other high-dimensional sparse domains, where Naive Bayes remains competitive with (or superior to) far more sophisticated learners that try to model feature dependence.


### 03. Gaussian Naive Bayes

**Question:**  
Gaussian Naive Bayes: for continuous features, model $P(x_i|y)$ as a Gaussian $\mathcal{N}(\mu_{y,i},\sigma_{y,i}^2)$.  
During training, estimate $\mu_y =$ mean of $x_i$ for class $y$, $\sigma_y^2 =$ variance of $x_i$ for class $y$.  
The likelihood becomes  
$$
P(\mathbf{x}|y) = \prod_i \frac{1}{\sqrt{2\pi\sigma_{y,i}^2}}\exp\Bigl(-\frac{(x_i-\mu_{y,i})^2}{2\sigma_{y,i}^2}\Bigr).

Compute this for $x=3.0$, $\mu_y=2.5$, $\sigma_y=1.0$.


**Answer:**

$$

$$
\frac{1}{\sqrt{2\pi}} \approx 0.39894, \qquad e^{-0.125} \approx 0.8825,
$$
$$

$$

### 04. Multinomial Naive Bayes

**Question:**  
Multinomial Naive Bayes: for count features (word frequencies in documents),  
model  
$$ P(\mathbf{x}|y) \propto \theta_y^{\mathbf{x}} \quad\text{where}\quad
\theta_{y,i} = \frac{\text{count of word } i \text{ in class } y + \alpha}{\text{total words in class } y + \alpha\cdot|V|}, $$  
with $\alpha=1$ (Laplace smoothing) and $V$ = vocabulary size.  

Explain why Laplace smoothing with $\alpha=1$ is essential — without it, a single unseen word zeros out the entire posterior.


**Answer:**

**Multinomial model**

Each class $y$ has a multinomial distribution over the vocabulary.  
The maximum-likelihood estimate of the probability of word $i$ under class $y$ is simply its relative frequency.  

**Laplace (add-one) smoothing**
$$
\hat\theta_{y,i} = \frac{N_{y,i}+\alpha}{N_y + \alpha\,|V|}, \qquad \alpha=1.
$$

**Why smoothing is mandatory**

Suppose a test document contains a word $w$ that never appeared in any training document of class $y$.  
Then the MLE $\hat\theta_{y,w}=0$, and therefore
$$
P(\mathbf{x}|y) = 0 \quad\Rightarrow\quad P(y|\mathbf{x})=0
$$
regardless of how well the rest of the document matches class $y$.  
A single unseen word would permanently eliminate that class.  

Adding a small pseudo-count $\alpha$ guarantees every word has positive probability, so the posterior remains well-defined and the classifier can still consider every class.


### 05. Bernoulli Naive Bayes

**Question:**  
Bernoulli Naive Bayes: for binary features (word present/absent),  
model $P(x_i=1|y)$ and $P(x_i=0|y)=1-P(x_i=1|y)$.  
Unlike Multinomial NB, Bernoulli NB **explicitly penalises the absence** of a word.  
Explain when Bernoulli NB outperforms Multinomial NB — typically for short documents where word presence matters more than frequency.


**Answer:**

**Bernoulli model**

Each feature is a binary indicator $x_i\in\{0,1\}$ (“word $i$ occurs at least once”).  
The class-conditional likelihood is
$$
P(\mathbf{x}|y) = \prod_{i=1}^{|V|} P(x_i|y)^{x_i}\,(1-P(x_i|y))^{1-x_i}.
$$
The factors $(1-P(x_i|y))$ for absent words are the key difference from the multinomial model.

**When Bernoulli wins**

- **Short documents / tweets / subject lines**: occurrence is far more informative than raw count; a word that appears once already carries the signal.  
- **Presence/absence is the natural feature**: e.g., medical symptom check-lists, binary sensor data.  

In longer documents the multinomial model (which can exploit repeated words) usually performs better.


### 06. Log-Space Computation

**Question:**  
Log-space computation: Naive Bayes multiplies many small probabilities, causing numerical underflow for long documents.  
The fix: compute  
$$
\log P(y|\mathbf{x}) \propto \log P(y) + \sum_i \log P(x_i|y).
$$
Show this is monotone in $P(y|\mathbf{x})$ so $\arg\max$ is preserved.  
Demonstrate underflow for a 100-word document where each $P(x_i|y)=0.01$ — compute the raw product and the log-sum.


**Answer:**

**Log-space identity**
$$
\log\Bigl(P(y)\prod_i P(x_i|y)\Bigr)
= \log P(y) + \sum_i \log P(x_i|y).
$$
Because $\log$ is strictly increasing, 
$$
\arg\max_y\, P(y|\mathbf{x}) = \arg\max_y\, \log P(y|\mathbf{x}).
$$
Hence the decision is unchanged.

**Numerical illustration**

Suppose a document of 100 independent tokens each with $P(x_i|y)=0.01$:


```python
import math

# Raw product – will underflow to 0.0 in floating point
raw = 0.01 ** 100
print(f"Raw product (0.01)**100 = {raw}")

# Log-sum – perfectly stable
log_sum = 100 * math.log(0.01)
print(f"log-sum = {log_sum:.4f}")
print(f"exp(log-sum) recovers the true value ≈ {math.exp(log_sum):.3e}")
```

### 01. Diagnosing a Spam-Filter Failure

**Question:**  
A spam filter trained with Naive Bayes classifies a genuine email containing the word “bank” as spam with $P(\text{Spam}|“bank”)=0.94$.  
The user complains. Diagnose the problem: what prior, likelihood, or data issue could cause this?  
Propose two specific fixes — one at the data level and one at the model level.


**Answer:**

**Possible causes**

1. **Biased training data** – the word “bank” appeared almost exclusively in spam (phishing) messages; legitimate financial emails were underrepresented.  
2. **Over-strong likelihood** – $P(\text{“bank”}|\text{Spam})$ was estimated far higher than $P(\text{“bank”}|\text{Ham})$ because of the skewed counts.  
3. **Prior mismatch** – the training prior $P(\text{Spam})$ may be higher than the user’s true spam rate.

**Two concrete fixes**

| Level   | Fix |
|---------|-----|
| **Data** | Add more legitimate financial / banking emails to the training set (or re-weight existing ones) so that the empirical $P(\text{“bank”}|\text{Ham})$ rises. |
| **Model** | Apply stronger smoothing / feature selection, or move to a discriminative classifier (e.g. Logistic Regression / linear SVM) that can down-weight the single feature “bank” when other features point to ham. |


### 02. Effect of Violating the Independence Assumption

**Question:**  
The naive independence assumption says that, given the class “Spam”, the words “free” and “win” are independent.  
In reality they often co-occur in the same emails.  
Explain whether this violation makes Naive Bayes overconfident, underconfident, or neither — and in which direction the posterior probability is biased.


**Answer:**

When two features are **positively dependent** given the class (they tend to appear together), Naive Bayes **double-counts** the evidence.

- The joint likelihood $P(\text{free},\text{win}|S)$ is approximated by the product $P(\text{free}|S)P(\text{win}|S)$, which is larger than the true joint probability.  
- Consequently the posterior $P(S|\text{free},\text{win})$ is **pushed higher** than it should be → **overconfidence**.

(The opposite occurs for negatively dependent features.)  
In short documents the bias is often tolerable; in long documents with many correlated words it can become severe, which is one reason calibrated probabilities from Naive Bayes are rarely trusted for decision-making under uncertainty.


### 03. Naive Bayes vs Logistic Regression

**Question:**  
Compare Naive Bayes to Logistic Regression on four dimensions in a table:  
(a) generative vs discriminative,  
(b) training data required,  
(c) feature independence assumption,  
(d) calibrated probabilities.  

Explain in which situation you would choose Naive Bayes over Logistic Regression despite the naive assumption.


**Answer:**

| Dimension                    | Naive Bayes                          | Logistic Regression                     |
|-----------------------------|--------------------------------------|-----------------------------------------|
| (a) Generative / Discriminative | **Generative** (models $P(x|y)P(y)$) | **Discriminative** (models $P(y|x)$ directly) |
| (b) Training data needed    | Works well with **small** data       | Needs more data to estimate decision boundary reliably |
| (c) Feature independence    | **Strong** independence assumption   | No independence assumption              |
| (d) Probability calibration | Often **poorly calibrated**          | Better calibrated (especially with regularisation) |

**When to prefer Naive Bayes**

- Extremely high-dimensional sparse data (text, bag-of-words) with limited labelled examples.  
- Need for very fast training and prediction.  
- The ranking of classes is more important than well-calibrated probabilities.  
- A strong, simple baseline is required before investing in more complex models.


### 04. Exact-Zero Posterior and Laplace Smoothing

**Question:**  
A student computes $P(\text{Spam}|\text{email})=0.000000000$ (exactly zero) for a new email.  
Explain exactly what caused this, why the model is nevertheless “confident”, and how Laplace smoothing fixes it.  
What value of $\alpha$ should you use and why?


**Answer:**

**Cause of the exact zero**

At least one word in the test email never appeared in any spam training document.  
Under maximum-likelihood estimation that word receives probability 0 under the spam class, making the whole product
$$
P(\text{email}|\text{Spam}) = 0 \quad\Rightarrow\quad P(\text{Spam}|\text{email})=0.
$$
(The model is “confident” in the mathematical sense that the posterior is exactly zero; the confidence is of course completely unwarranted.)

**How Laplace smoothing cures it**

By adding a pseudo-count $\alpha$ to every word count we guarantee
$$
\hat\theta_{w,\text{Spam}} > 0 \quad\text{for every word } w,
$$
so the product never collapses to zero.

**Recommended $\alpha$**

- Classic choice: $\alpha=1$ (add-one / Laplace smoothing).  
- In modern practice a smaller value such as $\alpha=0.1$ or even $\alpha=0.01$ is often preferred for large vocabularies, because $\alpha=1$ can over-smooth rare words.  
- The optimal $\alpha$ can be chosen by cross-validation on a held-out set.
