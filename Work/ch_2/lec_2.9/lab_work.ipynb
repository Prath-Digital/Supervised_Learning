# Lab Work - 2.9


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split,learning_curve,cross_val_score

np.random.seed(42)
```

# Q1 Model Complexity Comparison

```python
X=np.linspace(-3,3,100).reshape(-1,1)
y=np.sin(X).ravel()+np.random.normal(0,0.2,100)

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)

degrees=[1,3,10]

plt.figure(figsize=(12,6))

for d in degrees:
 model=make_pipeline(PolynomialFeatures(d),LinearRegression())
 model.fit(X_train,y_train)

 x_plot=np.linspace(-3,3,300).reshape(-1,1)
 y_plot=model.predict(x_plot)

 plt.scatter(X,y,s=10)
 plt.plot(x_plot,y_plot,label=f'Degree {d}')

plt.legend()
plt.show()
```

# Q2 Train vs Test Error

```python
train=[]
test=[]

deg=range(1,15)

for d in deg:
 model=make_pipeline(PolynomialFeatures(d),LinearRegression())
 model.fit(X_train,y_train)

train.append(mean_squared_error(y_train,model.predict(X_train)))
test.append(mean_squared_error(y_test,model.predict(X_test)))

plt.plot(deg,train,label='Train Error')
plt.plot(deg,test,label='Test Error')
plt.xlabel('Complexity')
plt.ylabel('Error')
plt.legend()
plt.show()
```

# Q3 Learning Curves

```python
model=make_pipeline(PolynomialFeatures(5),LinearRegression())

sizes,train_scores,test_scores=learning_curve(
model,
X,
y,
cv=5,
train_sizes=np.linspace(.1,1,10),
scoring='neg_mean_squared_error'
)

train_mean=-train_scores.mean(axis=1)
test_mean=-test_scores.mean(axis=1)

plt.plot(sizes,train_mean,label='Training')
plt.plot(sizes,test_mean,label='Validation')
plt.legend()
plt.show()
```

# Q4 Noise Impact

```python
noise=[0.1,1]

for n in noise:

 y2=np.sin(X).ravel()+np.random.normal(0,n,100)

 xtr,xts,ytr,yts=train_test_split(X,y2,test_size=.2)

 model=make_pipeline(PolynomialFeatures(5),LinearRegression())
 model.fit(xtr,ytr)

 train_err=mean_squared_error(ytr,model.predict(xtr))
 test_err=mean_squared_error(yts,model.predict(xts))

 print('Noise=',n)
 print('Train:',train_err)
 print('Test:',test_err)
```

# Q5 Underfitting Example

```python
model=make_pipeline(PolynomialFeatures(1),LinearRegression())
model.fit(X_train,y_train)

x=np.linspace(-3,3,300).reshape(-1,1)

plt.scatter(X,y)
plt.plot(x,model.predict(x))
plt.show()

print('Train Error=',mean_squared_error(y_train,model.predict(X_train)))
```

# Q6 Overfitting Example

```python
model=make_pipeline(PolynomialFeatures(12),LinearRegression())
model.fit(X_train,y_train)

plt.scatter(X,y)
plt.plot(x,model.predict(x))
plt.show()

print('Train=',mean_squared_error(y_train,model.predict(X_train)))
print('Test=',mean_squared_error(y_test,model.predict(X_test)))
```

# Q7 Regularization Effect

```python
ridge=make_pipeline(PolynomialFeatures(12),Ridge(alpha=1))
lasso=make_pipeline(PolynomialFeatures(12),Lasso(alpha=.01,max_iter=10000))

ridge.fit(X_train,y_train)
lasso.fit(X_train,y_train)

print('Ridge Test=',mean_squared_error(y_test,ridge.predict(X_test)))
print('Lasso Test=',mean_squared_error(y_test,lasso.predict(X_test)))
```

# Q8 Cross Validation

```python
model=make_pipeline(PolynomialFeatures(5),LinearRegression())

scores=cross_val_score(model,X,y,cv=5,
scoring='neg_mean_squared_error')

print('CV Scores:',-scores)
print('Average:',-scores.mean())
```
