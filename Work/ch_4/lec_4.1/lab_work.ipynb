# Q1 Ridge Regression (L2) — Manual Computation

```python
import numpy as np

# Dataset
x = np.array([1,2,3,4,5])
y = np.array([2,4,5,4,5])

lam = 1

# Design Matrix
X = np.column_stack((np.ones(len(x)), x))

print('X =')
print(X)

print('\ny =')
print(y)
```

```python
# OLS Estimate
beta_ols = np.linalg.inv(X.T @ X) @ (X.T @ y)

print('OLS Coefficients:')
print(beta_ols)

beta0_ols = beta_ols[0]
beta1_ols = beta_ols[1]

print(f'OLS Equation: y = {beta0_ols:.4f} + {beta1_ols:.4f}x')
```

```python
# Ridge Normal Equation
I = np.eye(X.shape[1])

ridge_matrix = X.T @ X + lam * I

print('X^T X =')
print(X.T @ X)

print('\nX^T y =')
print(X.T @ y)

print('\n(X^T X + λI) =')
print(ridge_matrix)
```

```python
# Ridge Solution
beta_ridge = np.linalg.inv(ridge_matrix) @ (X.T @ y)

print('Ridge Coefficients:')
print(beta_ridge)

beta0_ridge = beta_ridge[0]
beta1_ridge = beta_ridge[1]

print(f'Ridge Equation: y = {beta0_ridge:.4f} + {beta1_ridge:.4f}x')
```

### Comparison

Observe that Ridge shrinks the coefficients toward zero.
The magnitude of the slope becomes smaller compared to OLS.

# Q2 Lasso Regression (L1) — Coordinate Descent

```python
import numpy as np

x = np.array([1,2,3,4,5], dtype=float)
y = np.array([2,4,5,4,5], dtype=float)

lam = 1

# Center data
x_mean = np.mean(x)
y_mean = np.mean(y)

xc = x - x_mean
yc = y - y_mean

beta = 0.0

history = []
```

```python
def soft_threshold(rho, lam):
    if rho > lam:
        return rho - lam
    elif rho < -lam:
        return rho + lam
    else:
        return 0

for iteration in range(2):
    rho = np.sum(xc * yc)
    z = np.sum(xc**2)

    beta = soft_threshold(rho, lam) / z
    history.append(beta)

    print(f'Pass {iteration+1}: beta = {beta:.6f}')
```

```python
beta0_lasso = y_mean - beta*x_mean

print('Intercept:', beta0_lasso)
print('Slope:', beta)

print(f'Lasso Equation: y = {beta0_lasso:.4f} + {beta:.4f}x')
```

### Why Lasso can produce zeros

The soft-thresholding operator directly sets coefficients to zero when
their absolute contribution is smaller than λ.

# Q3 Visualization & Comparison

```python
import matplotlib.pyplot as plt

x_line = np.linspace(1,5,100)

y_ols = beta0_ols + beta1_ols*x_line
y_ridge = beta0_ridge + beta1_ridge*x_line
y_lasso = beta0_lasso + beta*x_line

plt.figure(figsize=(10,6))

plt.scatter(x,y,s=100,label='Data Points')

for i in range(len(x)):
    plt.text(x[i],y[i],f'({x[i]}, {y[i]})')

plt.plot(x_line,y_ols,label='OLS')
plt.plot(x_line,y_ridge,label='Ridge (L2)')
plt.plot(x_line,y_lasso,label='Lasso (L1)')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('OLS vs Ridge vs Lasso')
plt.legend()
plt.grid(True)
plt.show()
```

```python
lambdas = np.arange(0,11)

ridge_coeffs = []

for l in lambdas:
    B = np.linalg.inv(X.T@X + l*np.eye(2)) @ (X.T@y)
    ridge_coeffs.append(B[1])

plt.figure(figsize=(10,6))
plt.plot(lambdas,ridge_coeffs,marker='o')
plt.xlabel('Lambda')
plt.ylabel('Coefficient Value')
plt.title('Ridge Regularization Path')
plt.grid(True)
plt.show()
```

# Q4 Theory Questions

## Q4.1 Core Difference Between Ridge and Lasso

Ridge uses L2 penalty:

Loss = RSS + λΣβ²

Lasso uses L1 penalty:

Loss = RSS + λΣ|β|

Ridge shrinks coefficients continuously.
Lasso shrinks and can eliminate coefficients entirely.

## Q4.2 Why Lasso Produces Sparse Solutions

Geometrically:

- Ridge constraint region is circular.
- Lasso constraint region is diamond-shaped.
- Diamond corners occur on axes.
- Optimization frequently hits corners.
- Therefore coefficients become exactly zero.

## Q4.3 What Happens When λ → ∞

Ridge:
- Coefficients approach zero.
- Rarely become exactly zero.

Lasso:
- Coefficients become exactly zero.
- Model becomes extremely simple.
- Possible underfitting.

## Q4.4 When to Use Ridge

1. Multicollinearity among predictors.
2. When every feature may contain useful information.

Examples:
- House price prediction.
- Sales forecasting.
- Economic indicators.
- Sensor measurements.

## Q4.5 When to Use Lasso

1. Feature selection required.
2. High-dimensional datasets.

Examples:
- Gene expression analysis.
- Text classification.
- Customer churn prediction.
- Marketing analytics.

# Conclusion

- OLS provides the baseline model.
- Ridge reduces coefficient magnitude.
- Lasso can remove features entirely.
- Ridge handles multicollinearity well.
- Lasso performs automatic feature selection.
- Choice depends on the problem requirements.
